```python
import numpy as np
import pandas as pd
```

```python
np.random.seed(42)
```

```python
mu_vec1 = np.array([0,0,0])
cov_mat1 = np.array([[1,0,0],[0,1,0],[0,0,1]])
```

```python
class_samples = np.random.multivariate_normal(mu_vec1, cov_mat1, 20)
```

```python
class_samples
```

```python
df = pd.DataFrame(class_samples, columns=['X1', 'X2', 'X3'])
df['target'] = 1
df
```

```python
mu_vec2 = np.array([1,1,1])
cov_mat2 = np.array([[1,0,0],[0,1,0],[0,0,1]])
class2_samples = np.random.multivariate_normal(mu_vec2, cov_mat2, 20)
```

```python
class2_samples
df2 = pd.DataFrame(class2_samples, columns=['X1', 'X2', 'X3'])
df2['target'] = 0
```

```python
df = pd.concat([df, df2], ignore_index=True)
```

```python
df.sample(4)
```

```python
import plotly.express as px
```

```python
fig = px.scatter_3d(df, x='X1', y='X2', z='X3', color='target')
fig.update_traces(marker=dict(size=5, line=dict(width=2, color='DarkSlateGrey')), selector=dict(mode='markers'))
fig.show()
```
