```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer,MissingIndicator
```

```python
df = pd.read_csv('titanic/train.csv',usecols=['Age','Fare','Survived'])
```

```python
df.head()  
```

```python
X = df.drop('Survived',axis=1)
y = df['Survived']
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=0)
```

```python
X_train.head()
```

```python
X_test.head()
```

```python
si = SimpleImputer()
X_train_tr = si.fit_transform(X_train)
X_test_tr = si.transform(X_test)
```

```python
from sklearn.linear_model import LogisticRegression

lr = LogisticRegression()
lr.fit(X_train_tr, y_train)

y_pred = lr.predict(X_test_tr)

from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", accuracy)
```

```python
mi = MissingIndicator()

mi.fit(X_train)

X_train_mi = mi.transform(X_train)
X_test_mi = mi.transform(X_test)
```

```python
X_train_mi
```

```python
X_train['Age_missing'] = X_train_mi
X_test['Age_missing'] = X_test_mi
```

```python
X_train
```

```python
si = SimpleImputer()
X_train_tr2 = si.fit_transform(X_train)
X_test_tr2 = si.transform(X_test)
```

```python
lr = LogisticRegression()
lr.fit(X_train_tr2, y_train)

y_pred = lr.predict(X_test_tr2)

accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", accuracy)
```
