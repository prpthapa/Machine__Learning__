```python
from sklearn.datasets import make_regression
import matplotlib.pyplot as plt
import numpy as np
```

```python
X, y = make_regression(n_samples=4, n_features=1,n_informative=1, noise=80, random_state=42)
```

```python
plt.scatter(X,y)
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(X,y)
```

```python
model.coef_
```

```python
model.intercept_
```

```python
plt.scatter(X, y)
plt.plot(X, model.predict(X), color='red')
```

```python
# gradient descent

b = 0
m = -57.99
l = 0.01

epochs = 100
for i in range(epochs):
    loss_slope = -2 * np.sum(y - m*X.ravel() - b)
    b = b - l * loss_slope

    y_pred = m * X + b
    plt.plot(X, y_pred)

plt.scatter(X, y)
```
